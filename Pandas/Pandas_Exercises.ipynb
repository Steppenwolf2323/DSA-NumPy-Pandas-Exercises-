#### 2. Now print the first 5 and the last 5 rows of the dataset.

```python
# First approach: manual iteration over the file

for ind,rows in enumerate(dataframe.values):
    if ind < 5:
        print(rows)
print()

for rows in dataframe.values[-5:]:
    print(rows)

# Second approach:  Pandas methods

dataframe.head(5)
dataframe.tail(5)
```

#### 4. Find the most expensive car and print its name and price.

```python
display(dataframe[dataframe['price'] == dataframe['price'].max()])

dataframe[dataframe['price'] == dataframe['price'].max()][['company','price']]
```

#### 5. Print only the rows of a specific car company.

```python
# First Option: boolean filtering
display(dataframe[dataframe['company'] == 'volvo'])
print()

# Second Option: Pandas methods
volvo_cars = dataframe.groupby('company').get_group('volvo')
volvo_cars

# extra condition
new_data = pd.DataFrame(volvo_cars)
display(new_data[new_data['body-style'] == 'sedan'])
```

#### 5. Some practicality: give example of usage of .at, .iat, .loc and .iloc.

```python
print("The car and its price in the 3rd row is --> ", dataframe.at[3,"company"], dataframe.at[3,'body-style'].capitalize())
print()
print("The value in the 23th row and 7th column is --> ", dataframe.iat[23,7])
print()
print("The first 5 cars are: ")
display(dataframe.loc[0:4 , ["company"]])
print("The first 3 cars and their last 4 info are:")
display(dataframe.iloc[0:3, -4:])
```
