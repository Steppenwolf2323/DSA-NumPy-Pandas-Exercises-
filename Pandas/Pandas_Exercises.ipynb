#### 1. Let's start with a simple exercise. Read the dataset 'Automobile_data.csv' and report the number of rows and the title of the different columns.

```python
dataframe = pd.read_csv('Automobile_data.csv')

print(len(dataframe))
titles = list(dataframe.columns)
print(titles)
```

#### 2. Now print the first 5 and the last 5 rows of the dataset.

```python
# first way, the classical

for ind,rows in enumerate(dataframe.values):
    if ind < 5:
        print(rows)
print(f"\n")

for rows in dataframe.values[-5:]:
    print(rows)
```
